# Feature Engineering

## 1. Codificar categóricas

### 1.1 Ethnicity
En el dataset ya viene representado en un formato numérico como se muestra a continuación:
- 0 -> Caucasian
- 1 -> African
- 2 -> Asian
- 3 -> Other

Sin embargo, al ser datos categóricos dejarlos así podría causar en el modelo errores de predicción al creer que sí un paciente es *African* es menor que si es *Asian* dado que 1<2, es decir esta característica/variable no tiene un orden claro. Por lo tanto, se necesita transformar estos datos categóricos en un formato que los modelos puedan entender, para este caso se usó **One-Hot Encoding** la cual permite tratar cada categoría de forma independiente.

```python
import pandas as pd

alzheimer_data = pd.read_csv('../data/processed/alzheimers_disease_data.csv')

df = pd.get_dummies(alzheimer_data, columns= ['Ethnicity'], dtype= int)

# Muestra todas las columnas cuyo nombre contiene 'Ethnicity'
print(df.filter(like='Ethnicity').columns.tolist())

# Imprime solo las columnas que fueron generadas
print(df[['Ethnicity_0','Ethnicity_1', 'Ethnicity_2', 'Ethnicity_3']])
```

### 1.2 EducationLevel

En el dataset también ya viene representado numéricamente como se muestra a continuación:

- 0: None
- 1: High School
- 2: Bachelor's
- 3: Higher

A diferencia de la variable *Ethnicity*, estos datos categóricos sí tienen un orden implícito, existe una jerarquía clara desde no contar con ningún grado de estudios hasta haber realizado estudios universitarios. Por lo tanto, esta variable debe no codificarse de forma ordinal.

## Escalar variables numéricas

Variables como CholesterolTotal o SystolicBP tienen magnitudes superiores a 100, mientras que MMSE o BMI manejan rangos mucho más bajos. Esta diferencia de escalas puede hacer que las variables con valores más grandes dominen el entrenamiento y afecten las predicciones del modelo. Para evitar esto se usará StandardScaler, el cual estandariza las variables transformándolas para que tengan una media de 0 y una desviación estándar de 1.


```python
from sklearn.preprocessing import StandardScaler

numeric_columns = [
    'Age', 'BMI', 'AlcoholConsumption', 'PhysicalActivity', 'DietQuality',
    'SleepQuality', 'SystolicBP', 'DiastolicBP', 'CholesterolTotal',
    'CholesterolLDL', 'CholesterolHDL', 'CholesterolTriglycerides',
    'MMSE', 'FunctionalAssessment', 'ADL'
]

scaler = StandardScaler()
df[numeric_columns] = scaler.fit_transform(df[numeric_columns])

print(df[numeric_columns])
```
