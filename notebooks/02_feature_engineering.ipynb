## 1. Codificar categóricas

### 1.1 Ethnicity
En el dataset ya viene representado en un formato numérico como se muestra a continuación:
- 0 -> Caucasian
- 1 -> African
- 2 -> Asian
- 3 -> Other

Sin embargo, al ser datos categóricos dejarlos así podría causar en el modelo errores de predicción al creer que sí un paciente es *African* es menor que si es *Asian* dado que 1<2, es decir esta característica/variable no tiene un orden claro. Por lo tanto, se necesita transformar estos datos categóricos en un formato que los modelos puedan entender, para este caso se usó **One-Hot Encoding** la cual permite tratar cada categoría de forma independiente.

```python
import sys
sys.path.append('..')

import pandas as pd
from src.features import encode_categorical, split_features_target

alzheimer_data = pd.read_csv('../data/processed/alzheimers_disease_data.csv')

df = encode_categorical(alzheimer_data)
print(df.filter(like='Ethnicity').columns.tolist())
```

### 1.2 EducationLevel

En el dataset también ya viene representado numéricamente como se muestra a continuación:

- 0: None
- 1: High School
- 2: Bachelor's
- 3: Higher

A diferencia de la variable *Ethnicity*, estos datos categóricos sí tienen un orden implícito, existe una jerarquía clara desde no contar con ningún grado de estudios hasta haber realizado estudios universitarios. Por lo tanto, esta variable debe no codificarse de forma ordinal.


## 2. Definir X e Y

En el aprendizaje supervisado es fundamental definir *X* y *y*.

- **_X(Matriz de características o features):_**  Representa toda la información de entrada disponible del paciente (edad, nivel de colesterol, índice de masa corporal, resultados de pruebas cognitivas, hábitos de vida, etc.). Son las causas o indicadores.
- **_y(Vector Objetivo o Target):_** Representa el resultado exacto que el modelo debe aprender a diagnosticar o predecir (por ejemplo, si el paciente tiene o no Alzheimer: 1 o 0). Es la consecuencia o etiqueta.

```python
X, y = split_features_target(df, target_col='Diagnosis')

print(X)
print(y)

X.to_csv('../data/processed/X_features.csv', index=False)
y.to_csv('../data/processed/y_target.csv', index=False)
```
